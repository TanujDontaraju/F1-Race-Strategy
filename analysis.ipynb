# F1 Race Simulation Analysis

This notebook loads the data from `race_simulation_output.csv` and visualizes the results.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set a nice style for the plots
sns.set_theme(style="whitegrid")
```

### 1. Load the Simulation Data

```python
df = pd.read_csv('race_simulation_output.csv')
df.head()
```

### 2. Visualize Lap Times

```python
plt.figure(figsize=(16, 8))

# Plot the lap times
sns.lineplot(data=df, x='lap_number', y='lap_time', marker='o', label='Lap Time')

# Highlight the pit stop
pit_stop_lap = df[df['is_pit_stop'] == True]
if not pit_stop_lap.empty:
    pit_lap_number = pit_stop_lap['lap_number'].iloc[0]
    plt.axvline(x=pit_lap_number, color='red', linestyle='--', label=f'Pit Stop (Lap {pit_lap_number})')

plt.title('Simulated Race Lap Times', fontsize=16)
plt.xlabel('Lap Number')
plt.ylabel('Lap Time (seconds)')
plt.xticks(range(0, df['lap_number'].max() + 2, 2))
plt.legend()
plt.show()
```

### 3. Calculate Total Race Time

```python
total_race_time_seconds = df['lap_time'].sum()
minutes = int(total_race_time_seconds // 60)
seconds = total_race_time_seconds % 60

print(f"Total simulated race time: {total_race_time_seconds:.2f} seconds")
print(f"Which is {minutes} minutes and {seconds:.2f} seconds.")
```
